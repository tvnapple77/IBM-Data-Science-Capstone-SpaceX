# SpaceX Falcon 9 — EDA Visualization (six required figures)

**Data source:** `dataset_part_2.csv` from the IBM capstone lab (90 observations). Upload it into the same JupyterLab folder. This notebook generates the six requested visualizations and their summary tables. The 56-record Plotly Dash sample should not be mixed with this 90-record dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

path = Path("dataset_part_2.csv")
if not path.exists():
    raise FileNotFoundError("Upload dataset_part_2.csv to the same folder as this notebook")
df = pd.read_csv(path)
print("Rows, columns:", df.shape)
print("Columns:", list(df.columns))
assert {"FlightNumber", "LaunchSite", "PayloadMass", "Orbit", "Class"}.issubset(df.columns)
sns.set_theme(style="whitegrid")
```

## Figure 1: Flight Number vs Launch Site

```python
plt.figure(figsize=(11,5.8))
sns.scatterplot(data=df, x="FlightNumber", y="LaunchSite", hue="Class", palette={0:"#d65a56",1:"#237eaa"}, s=75)
plt.xlabel("Flight Number"); plt.ylabel("Launch Site")
plt.title("Flight Number vs Launch Site")
plt.tight_layout()
plt.savefig("spacex_eda_figure_1.png",dpi=180,bbox_inches="tight")
plt.show()
```

## Figure 2: Payload Mass vs Launch Site

```python
plt.figure(figsize=(11,5.8))
sns.scatterplot(data=df, x="PayloadMass", y="LaunchSite", hue="Class", palette={0:"#d65a56",1:"#237eaa"}, s=75)
plt.xlabel("Payload Mass (kg)");plt.ylabel("Launch Site")
plt.title("Payload Mass vs Launch Site")
plt.tight_layout()
plt.savefig("spacex_eda_figure_2.png",dpi=180,bbox_inches="tight")
plt.show()
```

## Figure 3: Landing Success Rate by Orbit

```python
plt.figure(figsize=(11,5.8))
orbit_stats = df.groupby("Orbit").agg(success_rate=("Class","mean"), launches=("Class","size")).sort_values("success_rate",ascending=False)
print(orbit_stats.to_string())
ax=sns.barplot(data=orbit_stats.reset_index(),x="Orbit",y="success_rate",color="#237eaa")
for i,r in enumerate(orbit_stats.itertuples()): ax.text(i,r.success_rate+.015,f"n={r.launches}",ha="center",fontsize=8)
plt.ylabel("Landing success rate");plt.ylim(0,1.12);plt.xticks(rotation=45,ha="right")
plt.title("Landing Success Rate by Orbit")
plt.tight_layout()
plt.savefig("spacex_eda_figure_3.png",dpi=180,bbox_inches="tight")
plt.show()
```

## Figure 4: Flight Number vs Orbit

```python
plt.figure(figsize=(11,5.8))
sns.scatterplot(data=df,x="FlightNumber",y="Orbit",hue="Class",palette={0:"#d65a56",1:"#237eaa"},s=75)
plt.title("Flight Number vs Orbit")
plt.tight_layout()
plt.savefig("spacex_eda_figure_4.png",dpi=180,bbox_inches="tight")
plt.show()
```

## Figure 5: Payload Mass vs Orbit

```python
plt.figure(figsize=(11,5.8))
sns.scatterplot(data=df,x="PayloadMass",y="Orbit",hue="Class",palette={0:"#d65a56",1:"#237eaa"},s=75)
plt.xlabel("Payload Mass (kg)")
plt.title("Payload Mass vs Orbit")
plt.tight_layout()
plt.savefig("spacex_eda_figure_5.png",dpi=180,bbox_inches="tight")
plt.show()
```

## Figure 6: Annual Landing Success Rate

```python
plt.figure(figsize=(11,5.8))
df["Year"] = pd.to_datetime(df["Date"]).dt.year
annual=df.groupby("Year").agg(success_rate=("Class","mean"),launches=("Class","size"))
print(annual.to_string())
plt.plot(annual.index,annual.success_rate,marker="o",linewidth=2)
plt.ylim(0,1.05);plt.ylabel("Landing success rate");plt.xlabel("Launch year")
plt.title("Annual Landing Success Rate")
plt.tight_layout()
plt.savefig("spacex_eda_figure_6.png",dpi=180,bbox_inches="tight")
plt.show()
```
